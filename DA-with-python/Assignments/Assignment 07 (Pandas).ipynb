1) Create a DataFrame from a dictionary with columns Name, Age, City and Salary for 6 employees. Print the first 3 rows, the shape, the column names and the data types.

```python
import numpy as np
import pandas as pd
```

```python
data1 = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [25, 34, 28, 45, 32, 29],
    "City": ["Delhi", "Mumbai", "Delhi", "Bangalore", "Delhi", "Mumbai"],
    "Salary": [35000, 52000, 48000, 75000, 42000, 38000],
}
df1 = pd.DataFrame(data1)


print("First 3 rows:\n", df1.head(3))
print("Shape:", df1.shape)
print("Column Names:", df1.columns.tolist())
print("Data Types:\n", df1.dtypes)
```

6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
df6 = df1.copy()

# Add bonus and total col
df6["Bonus"] = df6["Salary"] * 0.10
df6["Total"] = df6["Salary"] + df6["Bonus"]

# rename City to Loc
df6.rename(columns={"City": "Location"}, inplace=True)

# Drop Bonus col
df6.drop(columns=["Bonus"], inplace=True)

print("Updated DataFrame:\n", df6)
```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
df7 = pd.DataFrame(
    {
        "Name": ["Alice", "Bob", "Charlie", np.nan, "Eva"],
        "Age": [25, np.nan, 28, 45, 32],
        "Salary": [35000, 52000, np.nan, 75000, 42000],
    }
)

print("Original DataFrame with NaNs:\n", df7)
print("\nCount of missing values per column:\n", df7.isna().sum())

# Fill numeric NaNs with column mean
df7["Age"].fillna(df7["Age"].mean(), inplace=True)
df7["Salary"].fillna(df7["Salary"].mean(), inplace=True)

# Drop  missing values 
df7_cleaned = df7.dropna()

print("\nDataFrame after mean imputation and dropping remaining NaNs:\n", df7_cleaned)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
sorted_salary = df1.sort_values(by="Salary", ascending=False)
sorted_age = df1.sort_values(by="Age", ascending=True)
top_3_paid = df1.nlargest(3, "Salary")

print("Sorted by Salary (Descending):\n", sorted_salary)
print("\nSorted by Age (Ascending):\n", sorted_age)
print("\nTop 3 Highest-Paid Employees:\n", top_3_paid)
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
city_grouped = df1.groupby("City")["Salary"].agg(
    Avg_Salary="mean", Employee_Count="count", Max_Salary="max"
)

print("Groupby City Aggregations:\n", city_grouped)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
employees = pd.DataFrame(
    {
        "EmpID": [101, 102, 103, 104],
        "Name": ["Alice", "Bob", "Charlie", "David"],
        "DeptID": [1, 2, 1, 3],
    }
)

departments = pd.DataFrame(
    {
        "DeptID": [1, 2, 4],
        "DeptName": ["HR", "IT", "Finance"],
    }
)

inner_merged = pd.merge(employees, departments, on="DeptID", how="inner")
left_merged = pd.merge(employees, departments, on="DeptID", how="left")

# DataFrames for concatenation
df_part1 = pd.DataFrame(
    {"EmpID": [101, 102], "Name": ["Alice", "Bob"], "DeptID": [1, 2]}
)
df_part2 = pd.DataFrame(
    {"EmpID": [102, 103], "Name": ["Bob", "Charlie"], "DeptID": [2, 1]}
)

concatenated = pd.concat([df_part1, df_part2], ignore_index=True)
deduplicated = concatenated.drop_duplicates()

print("Inner Join:\n", inner_merged)
print("\nLeft Join:\n", left_merged)
print("\nConcatenated (with duplicates):\n", concatenated)
print("\nAfter Dropping Duplicates:\n", deduplicated)
```
