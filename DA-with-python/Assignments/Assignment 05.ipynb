QUESTION 1: Working with Lists
- Create a list called numbers containing at least 8 integers.
- Using a for loop, do the following:
- Print all numbers in the list.
- Print the sum of all numbers.
- Print only the even numbers

Requirements:
- Use at least one loop.
- Use conditional statements.
- Display clear output

```python
numbers = [12, 7, 19, 24, 30, 5, 8, 42]


total_sum = 0
even_numbers = []


print("Numbers in the list:")
# Using a for loop 
for num in numbers:
    print(num, end=" ")
    total_sum += num  #  sum
    if num % 2 == 0:  # Check for even numbers 
        even_numbers.append(num)

print()  # Newline for clean output
print(f"Sum of all numbers: {total_sum}")
print(f"Even numbers: {even_numbers}\n")
```

QUESTION 2: Dictionary Basics
Create a dictionary called student with the following keys:
- name
- age
- course
- marks (must store a list of 3 marks)
Using loops:
- Print all keys and their values.
- Calculate the average of the marks.
- Display whether the student Passed (average ≥ 50) or Failed.

Requirements:
- Dictionary must contain a list.
- Use loops to calculate the average.



```python
student = {
    "name": "Sarah Connor",
    "age": 20,
    "course": "Computer Science",
    "marks": [78, 85, 62]
}

# Print all keys 
print("Student Details:")
for key, value in student.items():
    print(f"  {key}: {value}")

#  the average marks 
total_marks = 0
for mark in student["marks"]:
    total_marks += mark

average_mark = total_marks / len(student["marks"])
print(f"Average Mark: {average_mark:.2f}")
# Check pass status 
if average_mark >= 50:
    print("Status: Passed\n")
else:
    print("Status: Failed\n")
```

QUESTION 3: List of Dictionaries
- Create a list called employees.
- Each employee must be stored as a dictionary containing:
- name
- department
- salary

Using loops:
- Print details of all employees.
- Find the employee with the highest salary.
- Calculate the total salary expense.

Requirements:
- Minimum 3 employees.
- Use comparison logic.
- Use loops to process data.



```python
employees = [
    {"name": "Alice Smith", "department": "HR", "salary": 55000},
    {"name": "Bob Johnson", "department": "IT", "salary": 82000},
    {"name": "Charlie Brown", "department": "Finance", "salary": 67000}
]

total_salary_expense = 0
highest_paid_employee = employees[0]

print("Employee Records:")
for emp in employees:
    # Print details
    print(f"  Name: {emp['name']}, Dept: {emp['department']}, Salary: ${emp['salary']}")
    
    # total salary 
    total_salary_expense += emp["salary"]
    
    # find the highest salary
    if emp["salary"] > highest_paid_employee["salary"]:
        highest_paid_employee = emp

print(f"\nTotal Salary Expense: ${total_salary_expense}")
print(f"Highest Paid Employee: {highest_paid_employee['name']} (${highest_paid_employee['salary']})\n")
```

QUESTION 4: While Loop with User Input
- Ask the user to enter numbers.
- Store the numbers in a list.
- Stop when the user enters -1.
- After the loop ends:
- Print the complete list.
- Print the largest number.
- Print the smallest number.

Requirements:
- Must use a while loop.
- Must store input in a list.
- Handle user input correctly.


```python
user_numbers = []

while True:
    user_input = int(input("Enter an integer (-1 to stop): "))
    if user_input == -1:
        break
    user_numbers.append(user_input)

# Handle output 
if len(user_numbers) > 0:
    # Finding largest and smallest numbers
    largest = user_numbers[0]
    smallest = user_numbers[0]
    for val in user_numbers:
        if val > largest:
            largest = val
        if val < smallest:
            smallest = val

    print(f"Complete list: {user_numbers}")
    print(f"Largest number: {largest}")
    print(f"Smallest number: {smallest}\n")
else:
    print("No numbers were entered.\n")
```

QUESTION 5: Word Frequency Counter
- Ask the user to enter a sentence.
- Convert the sentence into a list of words.
- Use a dictionary to count how many times each word appears.
- Print the word frequency clearly.
- Example Format (for understanding only):
apple: 3
banana: 2
orange: 1

Requirements:
- Use string splitting.
- Use a dictionary for counting.
- Use loops for processing.


```python
sentence = input("Enter a sentence: ")

words = sentence.split()


word_counts = {}
for word in words:
    
    clean_word = word.lower()
    if clean_word in word_counts:
        word_counts[clean_word] += 1
    else:
        word_counts[clean_word] = 1

print("\nWord Frequency:")
for word, count in word_counts.items():
    print(f"{word}: {count}")
print()
```
