6. Take the radius of a circle from a variable and calculate its Area and Circumference (use `pi = 3.14159`)

```python
radius = 5
pi = 3.14
Area = pi * radius ** 2 
Circumference = 2 * pi * radius
print("Radius of Circle is", radius)
print("Area of Circle is", Area)
print("Circumference of Circle is", Circumference)
```

7. Take the price of an item and a discount percentage in variables, then calculate and print the discount amount and the final price after discount

```python
price = int(input("Enter the price of the product: "))
discout_percentage = 15
discount_amount = price * discout_percentage / 100 
final_price = price - discount_amount
print("The original price of the product is:", price)
print("The discount amount is:", discount_amount)
print("The final price after discount is:", final_price)
```
