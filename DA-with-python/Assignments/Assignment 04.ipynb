Q3 — Loop Through Dictionary
person = {
"name": "Ali",
"age": 25,
"city": "Karachi"
}
Print keys and values using loop.

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}
for key, value in person.items():
    print(f"{key}: {value}")
```

Q4 — Count Frequency of Numbers
numbers = [1,2,2,3,3,3,4,4,4,4]


```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
freq = {}
for num in numbers:
    freq[num] = freq.get(num, 0) + 1
print(freq)
```

Q5 — Word Counter
Take sentence input and count word frequency

```python
sentence = input("Enter a sentence: ")
words = sentence.lower().split()
word_count = {}
for word in words:
    word_count[word] = word_count.get(word, 0) + 1
print(word_count)
```

Q6 — Student Marks & Average
marks = {"Math":80, "English":70, "Physics":90}
Calculate total and average.

```python
marks = {"Math": 80, "English": 70, "Physics": 90}
total_marks = sum(marks.values())
average_marks = total_marks / len(marks)
print(f"Total: {total_marks}")
print(f"Average: {average_marks:.2f}")
```

Q7 — Highest Scoring Student
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}
Find highest scoring student.


```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}
top_student = max(students, key=students.get)
print(f"Highest scorer: {top_student} with {students[top_student]} marks")
```

Q8 — Merge Dictionaries
dict1 = {"a":1,"b":2}
dict2 = {"c":3,"d":4}
Merge and print result

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
