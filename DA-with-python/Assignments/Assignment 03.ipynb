1. Write a program that accepts a list from user and print the alternate element of list.

```python
raw_input = input("Enter list elements separated by spaces: ")
elements = raw_input.split()

alternate_elements = elements[::2]
print("Alternate elements:", alternate_elements)
```

2. Write a program that accepts a list from user. Your program should reverse the content of list and 
display it. Do not use reverse() method.  

```python
raw_input = input("Enter list elements separated by spaces: ")
elements = raw_input.split()

reversed_list = elements[::-1]
print("Reversed list:", reversed_list)
```

3. Find and display the largest number of a list without using built-in function max(). Your program 
should ask the user to input values in list from keyboard. 

```python
raw_input = input("Enter numbers separated by spaces: ")
numbers = [float(x) for x in raw_input.split()]
if not numbers:
    print("The list is empty.")
else:
    largest = numbers[0]
    for num in numbers[1:]:
        if num > largest:
            largest = num

    display_val = int(largest) if largest.is_integer() else largest
    print("Largest number:", display_val)
```

4. Write a program that rotates the element of a list so that the element at the first index moves to the 
second index, the element in the second index moves to the third index, etc., and the element in the last 
index moves to the first index.

```python
raw_input = input("Enter list elements separated by spaces: ")
elements = raw_input.split()
if elements:
    
    rotated = [elements[-1]] + elements[:-1]
    print("Rotated list:", rotated)
else:
    print("List is empty.")
```

5. Write a program that input a string and ask user to delete a given word from a string.  

```python
sentence = input("Enter a sentence: ")
word = input("Enter the word you want to delete: ")
words = sentence.split()
new_sentence = ""
for w in words:
    if w != word:
        new_sentence += w + " "

print("New string:", new_sentence.strip())
```

6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It 
should print the date in the form March 12, 2021. 

```python
date_input = input("Enter date : ").strip()

month_names = [
    "", "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]
parts = date_input.split("/")
if len(parts) == 3:
    mm, dd, yyyy = int(parts[0]), int(parts[1]), parts[2]
    formatted_date = f"{month_names[mm]} {dd}, {yyyy}"
    print("Formatted date:", formatted_date)
else:
    print("Invalid format. Please use MM/DD/YYYY.")
```

7. Write a program with a function that accepts a string from keyboard and create a new string after 
converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." 
the output should be "Stop And Smell The Roses"

```python
def capitalize_words(sentence):
    words = sentence.split()
    new_sentence = ""
    for word in words:
        new_sentence += word.capitalize() + " "
    
    return new_sentence.strip()
sentence = input("Enter a sentence: ")
result = capitalize_words(sentence)
print(result)
```

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be 
like this :
 
2  11  7  12
5  2   9  15
8  3  10  42 

Sum of row 1 = 32 
Sum of row 2 = 31 
Sum of row 3 = 63

```python
matrix = [
    [2, 11, 7, 12],
    [5, 2, 9, 15],
    [8, 3, 10, 42]
]
for i in range(len(matrix)):
    total = 0

    for j in range(len(matrix[i])):
        total += matrix[i][j]

    print("Sum of row", i + 1, "=", total)
```

9. Write a program to add two matrices of size n x m.

```python
matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]
matrix2 = [
    [7, 8, 9],
    [10, 11, 12]
]
result = []
for i in range(len(matrix1)):
    row = []
    for j in range(len(matrix1[0])):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)
print("Result:")
for row in result:
    print(row)
```

10. Write a program to multiply two matrices 

```python
matrix1 = [
    [1, 2],
    [3, 4]
]
matrix2 = [
    [5, 6],
    [7, 8]
]
result = [
    [0, 0],
    [0, 0]
]
for i in range(len(matrix1)):
    for j in range(len(matrix2[0])):
        for k in range(len(matrix2)):
            result[i][j] += matrix1[i][k] * matrix2[k][j]
print("Result:")
for row in result:
    print(row)
```
